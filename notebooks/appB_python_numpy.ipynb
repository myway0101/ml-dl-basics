## 1. 파이썬 기본
### 값과 변수

```python
km = 12.9                 # 실수(float)
fare = 18900              # 정수(int)
name = "아델리"            # 문자열(str)
is_long = km > 10         # 참/거짓(bool)
print(type(km), type(fare), type(name), is_long)
```

### 리스트
번호(인덱스)는 0부터 셉니다. `kms[0:2]`는 0번부터 2번 **앞**까지입니다.

```python
kms = [2.0, 4.0, 6.0]       # 리스트: 대괄호로 묶는다
print(kms[0], kms[-1])       # 첫 원소, 마지막 원소
print(kms[0:2])              # 0번부터 2번 앞까지
print(len(kms))              # 원소의 개수
kms.append(8.0)              # 끝에 하나 더하기
print(kms)
```

### 튜플과 딕셔너리
딕셔너리는 값마다 이름(키)을 붙여 두고 이름으로 꺼냅니다. 케라스의 학습 기록(`hist.history`)이 딕셔너리입니다.

```python
point = (45.0, 215.0)        # 튜플: 바꿀 수 없는 묶음
bill, flipper = point        # 한 번에 나누어 담기
print(bill, flipper)

H = {"loss": [0.69, 0.41], "accuracy": [0.62, 0.88]}   # 딕셔너리
print(H["loss"])             # 이름(키)으로 값 꺼내기
print(len(H["loss"]))
```

### 되풀이와 조건
`range(시작, 끝, 간격)`, `enumerate()`(번호와 값), `zip()`(두 리스트를 나란히), `if`(조건).

```python
for k in range(1, 8, 2):            # 1부터 8 앞까지 2씩
    print(k, end=" ")
print()

species = ["Adelie", "Chinstrap", "Gentoo"]
counts = [151, 68, 123]
for i, s in enumerate(species, start=1):  # 번호와 값을 함께
    print(i, s)
for s, n in zip(species, counts):         # 두 리스트를 나란히
    if n > 100:
        print(f"{s}: {n}마리")
```

### 함수
7장의 기록 세 건으로 MSE를 계산하는 함수를 만듭니다.

```python
def mse(y, y_hat):                    # 함수 만들기
    total = 0
    for a, b in zip(y, y_hat):
        total += (a - b) ** 2              # ** 는 거듭제곱
    return total / len(y)

print(mse([7, 8, 12], [6.5, 9, 11.5]))   # 7장의 기록 세 건
```

### 리스트 내포와 f-문자열
| 형식 | 뜻 |
|---|---|
| `{x:.3f}` | 소수 셋째 자리까지 |
| `{x:.1%}` | 퍼센트로, 소수 첫째 자리까지 |
| `{x:,}` | 세 자리마다 쉼표 |
| `{x:2d}` | 정수를 두 칸에 맞춰 |

```python
squares = [x ** 2 for x in [1, 2, 3, 4]]      # 리스트 내포
print(squares)
print(sum(x > 2 for x in [1, 2, 3, 4]))      # 조건이 참인 개수

acc, rmse, n = 0.97222, 306.531, 1797
print(f"정확도 {acc:.1%}, RMSE {rmse:.1f} g, 이미지 {n:,}장")
print(f"[{3:2d}] [{12:2d}] 소수 셋째 자리: {2 / 3:.3f}")
```

## 2. 넘파이
### 리스트와 배열은 무엇이 다른가
리스트에 2를 곱하면 이어 붙이고, 배열에 2를 곱하면 원소마다 곱합니다.

```python
import numpy as np

lst = [1, 2, 3]
print(lst * 2)                  # 리스트: 통째로 이어 붙인다
print([v * 2 for v in lst])     # 리스트는 for로 하나씩
arr = np.array([1, 2, 3])
print(arr * 2)                  # 배열: 원소마다 곱한다
```

### 배열의 원소별 계산
같은 자리의 원소끼리 계산됩니다. 마지막 줄은 7장의 MSE를 한 줄로 쓴 것입니다.

```python
x = np.array([2.0, 4.0, 6.0])           # 1차원 배열
y = np.array([7.0, 8.0, 12.0])
print(x.shape)                           # 모양: 원소 3개
print(x * 2 + 1)                         # 원소마다 계산
print(y - x)                             # 같은 자리끼리 계산
print(np.mean((y - (1.25 * x + 4)) ** 2))   # 7장의 MSE 계산
```

### 2차원 배열과 인덱싱
대괄호 안에 **행 번호, 열 번호** 순서로 씁니다. `:`는 "그 방향 전부"입니다.

```python
X = np.array([[39.1, 181.0],
              [39.5, 186.0],
              [40.3, 195.0]])   # 2차원 배열: 행 = 샘플, 열 = 특징
print(X.shape)                  # (행 수, 열 수)
print(X[0])                     # 첫 행(첫 펭귄)
print(X[:, 1])                  # 모든 행의 둘째 열(날개 길이)
print(X[1, 0])                  # 둘째 행, 첫째 열
```

### 축을 따라 모으기: `axis`
- `axis=0`: 위아래로 모아 **열마다** 하나 (과목마다 평균)
- `axis=1`: 옆으로 모아 **행마다** 하나 (학생마다 평균)
- `axis`를 주지 않으면 전체를 하나로

```python
S = np.array([[90, 80],
              [70, 60],
              [85, 95]])     # 학생 3명(행) × 과목 2개(열)의 점수
print(S.mean(axis=0))        # axis=0: 위아래로 모아 과목마다 평균
print(S.mean(axis=1))        # axis=1: 옆으로 모아 학생마다 평균
print(S.sum(), S.max(axis=0), S[:, 1].argmax())
```

### 브로드캐스팅
`X`는 3×2, `X.mean(axis=0)`은 원소 2개입니다. 넘파이가 평균을 세 행으로 펼쳐 빼 줍니다(3장의 표준화).

```python
z = (X - X.mean(axis=0)) / X.std(axis=0)   # 열마다 표준화(3장)
print(z.round(2))
```

### 모양 바꾸기: `reshape(-1, 1)`
사이킷런은 입력 `X`가 (샘플 수, 특징 수) 모양의 2차원이기를 바랍니다. 특징이 하나뿐일 때 1차원 배열을 열 하나짜리 표로 세웁니다.

```python
km = np.array([3.0, 12.5, 25.0])
print(km.shape)                  # 1차원: (3,)
K = km.reshape(-1, 1)            # 열 하나짜리 2차원으로
print(K.shape)
print(K)
```

### 조건으로 고르고 세기
참·거짓 배열로 ① 0과 1로 바꾸기 ② 개수 세기 ③ 고르기를 할 수 있습니다.

```python
proba = np.array([0.12, 0.81, 0.47, 0.95])
print(proba >= 0.5)                      # 원소마다 참/거짓
print((proba >= 0.5).astype(int))        # 참은 1, 거짓은 0
print((proba >= 0.5).sum())              # 참의 개수
print(proba[proba >= 0.5])               # 조건에 맞는 원소만
```

### 난수와 수학 함수
시드를 정하면 난수가 매번 같게 나옵니다.

```python
rng = np.random.default_rng(seed=7)      # 시드를 정하면 결과가 같다
print(rng.uniform(1, 25, size=3).round(1))     # 1~25 사이 실수 3개
print(rng.choice(10, size=5, replace=False))  # 0~9에서 5개
print(np.exp(1.0).round(3), np.log(np.e))     # e, 자연로그
```

## 3. 판다스
### 불러오기와 열 고르기

```python
import seaborn as sns

df = sns.load_dataset("penguins")
cols = ["species", "bill_length_mm", "flipper_length_mm",
        "body_mass_g"]
print(df[cols].head(3))
```

대괄호 하나는 열 하나(1차원), 대괄호 둘은 열 목록으로 된 표(2차원)입니다. 사이킷런의 특징 `X`에는 2차원 표를 넣습니다.

```python
print(df["body_mass_g"].shape)      # 대괄호 하나: 열 하나(1차원)
print(df[["body_mass_g"]].shape)    # 대괄호 둘: 열 목록(2차원 표)
```

### 빈 칸 다루기
빈 칸이 섞인 네 줄짜리 표로 연습합니다. 판다스는 빈 칸을 `NaN`으로 표시합니다.

```python
import pandas as pd

t = pd.DataFrame({
    "species": ["Adelie", "Adelie", "Gentoo", "Gentoo"],
    "bill_length_mm": [39.1, np.nan, 46.1, 50.0],
    "sex": ["Male", "Female", np.nan, "Male"]})
print(t)
```

**빈 칸 찾기**: `isna()`는 칸마다 비었는지를, `sum()`은 열마다 빈 칸의 수를 알려 줍니다.

```python
print(t.isna())            # 칸마다 비었는가
print(t.isna().sum())      # 열마다 빈 칸의 수
```

**빈 칸이 있는 행 빼기**: `subset`으로 열을 정하면 그 열이 빈 행만 뺍니다.

```python
print(t.dropna())                          # 빈 칸 있는 행 빼기
print(t.dropna(subset=["bill_length_mm"]))  # 이 열만 보고
```

**빈 칸 채우기**: 빈 칸을 중앙값으로 채웁니다. 실제로는 채우는 값을 **학습 데이터로만** 계산해야 합니다(3장).

```python
med = t["bill_length_mm"].median()          # 빈 칸을 뺀 중앙값
print(med)
print(t["bill_length_mm"].fillna(med))      # 중앙값으로 채우기
```

실제 펭귄 데이터에서도 확인합니다.

```python
print(df["body_mass_g"].isna().sum())          # 빈 칸의 수
df = df.dropna(subset=["body_mass_g"])   # 빈 칸이 있는 행 빼기
print(df.shape)
print(df["species"].value_counts())            # 값마다 개수
```

### 묶어 계산하기와 행 고르기
`groupby()`는 나누기 → 계산하기 → 합치기를 한 번에 합니다. `iloc`은 위치로, `loc`은 이름으로 칸을 고릅니다.

```python
print(df.groupby("species")["body_mass_g"].mean().round(1))
heavy = df[df["body_mass_g"] > 5500]     # 조건으로 행 고르기
print(len(heavy), set(heavy["species"]))
print(df.iloc[0, 2], df.loc[0, "bill_length_mm"])  # 위치, 이름
```

```python
X = df[["bill_length_mm", "flipper_length_mm"]].values
print(type(X), X.shape)
print(X[:2])
```
